 Create a figure containing 4 plots 𝒚𝟏=𝒄𝒐𝒔(𝒙),  𝒚𝟐=𝒆−𝒙𝒄𝒐𝒔(𝒙), 𝒚𝟑= 𝟏/𝟏+𝒆−𝒙, 𝒚𝟓=𝒔𝒊𝒏(𝒙)/𝒙

```python
x = np.arange(-10, 10, 0.1)

y1 = np.cos(x)
y2 = np.exp(-x) * np.cos(x)
y3 = 1 / (1 + np.exp(-x))
y4 = np.sin(x)/ x

fig, axis = plt.subplots(2, 2)

axis[0,0].plot(x, y1)
axis[0,0].set_title('y1 = cos(x)')

axis[0,1].plot(x, y2)
axis[0,1].set_title('y2 = exp(-x)*cos(x)')

axis[1,0].plot(x, y3)
axis[1,0].set_title('y3 = 1/(1+exp(-x))')

axis[1,1].plot(x, y4)
axis[1,1].set_title('y4 = sin(x)/x')

plt.tight_layout()
plt.show()
```

generate 3D Mathematical Surface  𝒁=𝐬𝐢𝐧(√𝒙𝟐+𝒚𝟐)

```python
x = np.arange(-10, 10, 0.1)
y = np.arange(-10, 10, 0.1)
x, y = np.meshgrid(x, y)
z = np.sin(np.sqrt(x**2 + y**2))

fig = plt.figure()
ax = fig.add_subplot(projection='3d')
ax.plot_surface(x, y, z)

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')

plt.show()
```

 create a matrix 50x50 and visualize it using  𝑨𝒊𝒋=𝒔𝒊𝒏(𝒊)𝐜𝐨𝐬(𝐣) Then visualize it using Heatmap

```python
import seaborn as sns

A = np.zeros((50, 50))

for i in range(50):
    for j in range(50):
        A[i, j] = np.sin(i) * np.cos(j)

sns.heatmap(A)
plt.show()
```
